```python
# import necessary libraries
import requests
import json
import pandas as pd
import os
```

```python
# initialize the API URL
# the cursor allows us to pull multiple pages of results to get all SFU works in one call. 
# NOTE: SFU's institutional id in OpenAlex is I18014f758

# filtered only on short keyword phrase and institution (not years - though it shouldnt matter), and included xpac

url_with_cursor = 'https://api.openalex.org/works?search=("illicitly manufactured fentanyl")&filter=institutions.id:I18014758&include_xpac=true&per_page=200&cursor={}'
cursor = '*'


# empty containter to store the results
results_manufactured_fentanyl = []

# loop through pages
while cursor:
    
    # set cursor value and request page from OpenAlex
    url = url_with_cursor.format(cursor)
    
    #print("\n" + url)
    page_with_results = requests.get(url)
    page_with_results = page_with_results.json()
    
    # loop through partial list of results
    results = page_with_results['results']
    
    results_manufactured_fentanyl += results

    # update cursor to meta.next_cursor
    cursor = page_with_results['meta']['next_cursor']
    #print("next cursor is: ", cursor)


pd.DataFrame(results_manufactured_fentanyl).to_csv('./openalex_keyword_test/manufactured_fentanyl.csv', index = False)
```
