# OpenAlex query wildcard test cases

We noticed some unexpected behaviour with the OpenAlex query specifically to do with wildcards, so this workbook runs a suite of test cases to better understand the functionality

```python
# import necessary libraries
import requests
import json
import pandas as pd
import os
```

## manufacture/ing

```python
# filtered only on short keyword phrase and institution (not years - though it shouldnt matter), and included xpac

url_with_cursor = 'https://api.openalex.org/works?search=("illicitly manufactured fentanyl")&filter=institutions.id:I18014758&include_xpac=true&per_page=200&cursor={}'
cursor = '*'

results_manufactured_fentanyl = []

while cursor:
    url = url_with_cursor.format(cursor)
    page_with_results = requests.get(url)
    page_with_results = page_with_results.json()
    results = page_with_results['results']
    results_manufactured_fentanyl += results
    cursor = page_with_results['meta']['next_cursor']

pd.DataFrame(results_manufactured_fentanyl).to_csv('./openalex_keyword_test/rerun/manufactured_fentanyl.csv', index = False)
```

```python
url_with_cursor = 'https://api.openalex.org/works?search.exact=(manufactur*)&filter=institutions.id:I18014758&include_xpac=true&per_page=200&cursor={}'
cursor = '*'

results_manufactur = []

while cursor:
    url = url_with_cursor.format(cursor)
    page_with_results = requests.get(url)
    page_with_results = page_with_results.json()
    print(page_with_results)
    results = page_with_results['results']
    results_manufactur += results
    cursor = page_with_results['meta']['next_cursor']

pd.DataFrame(results_manufactur).to_csv('./openalex_keyword_test/rerun/manufactur.csv', index = False)
```

**rerun note:** originally the query had ("manufactur*") in it, and the API returned an error telling us not to use wildcards inside of a quoted phrase. Now we changed it to manufactur* and the rerun gave us a new error telling us to use exact search, with keyword search.exact. After implementing this,we see that the full results come back as intended 

```python
url_with_cursor = 'https://api.openalex.org/works?search=(manufacturing)&filter=institutions.id:I18014758&include_xpac=true&per_page=200&cursor={}'
cursor = '*'

results_manufacturing = []

while cursor:
    url = url_with_cursor.format(cursor)
    page_with_results = requests.get(url)
    page_with_results = page_with_results.json()
    results = page_with_results['results']
    results_manufacturing += results
    cursor = page_with_results['meta']['next_cursor']

pd.DataFrame(results_manufacturing).to_csv('./openalex_keyword_test/rerun/manufacturing.csv', index = False)
```
